```python
# ============ TOPIC 1A: CONDITIONAL STATEMENTS ============# ---------- Variables ----------
student_name = "Leticia Namutebi"
score = 72

# ---------- if / elif / else ----------
if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"
print(f"{student_name} scored {score} and got grade {grade}")

# ---------- Nested conditions ----------
age = 20
has_id = True
if age >= 18:
    if has_id:
        print("Access granted")
    else:
        print("Please bring your ID")
else:
    print("Too young to enter")

# ---------- Combining conditions (and / or / not) ----------
temperature = 28
is_raining = False
if temperature > 25 and not is_raining:
    print("Good day for a walk")
```

```python
# ============ TOPIC 1B: LOOPS ============

# ---------- for loop over a list ----------
fruits = ["apple", "mango", "banana"]
for fruit in fruits:
    print(fruit)

# ---------- for loop with range() ----------
for i in range(1, 6):
    print("Square of", i, "is", i ** 2)

# ---------- enumerate(): index + value ----------
for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)

# ---------- while loop ----------
count = 1
while count <= 5:
    print("Count:", count)
    count += 1

# ---------- break and continue ----------
for n in range(1, 11):
    if n == 3:
        continue      # skip 3
    if n == 8:
        break         # stop the loop at 8
    print(n)

# ---------- Nested loop: multiplication table ----------
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end="\t")
    print()

# ---------- Data-science style example ----------
salaries = [1200, 1500, 900, 2000]
total = 0
for s in salaries:
    total += s
print("Total:", total, "| Average:", total / len(salaries))
```

```python
# ============ TOPIC 2A: FUNCTIONS ============

# ---------- Basic function with a docstring ----------
def greet(name):
    """Return a greeting for the given name."""
    return f"Hello, {name}!"

print(greet("Leticia"))

# ---------- Default arguments ----------
def power(base, exponent=2):
    """Raise base to exponent (default is square)."""
    return base ** exponent

print(power(5))       # 25
print(power(2, 10))   # 1024

# ---------- Returning multiple values ----------
def describe(numbers):
    """Return min, max and mean of a list."""
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, mean = describe([4, 8, 15, 16, 23, 42])
print("Min:", low, "Max:", high, "Mean:", round(mean, 2))

# ---------- *args and **kwargs ----------
def add_all(*args):
    return sum(args)

def show_info(**kwargs):
    for key, value in kwargs.items():
        print(key, "=", value)

print(add_all(1, 2, 3, 4))
show_info(name="Leticia", course="Python", year=2026)

# ---------- Modular programming: small functions combined ----------
def clean_scores(scores):
    return [s for s in scores if 0 <= s <= 100]

def average(scores):
    return sum(scores) / len(scores)

raw = [45, 88, 105, 67, -3, 92]
print("Average of valid scores:", average(clean_scores(raw)))
```

```python
# ============ TOPIC 2B: RECURSION ============

def factorial(n):
    """Recursive factorial: n! = n * (n-1)!"""
    if n == 0 or n == 1:       # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print("5! =", factorial(5))    # 120

def fibonacci(n):
    """Return the nth Fibonacci number recursively."""
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(i) for i in range(10)])

# ============ TOPIC 2C: LAMBDA FUNCTIONS ============

square = lambda x: x ** 2
print(square(7))

add = lambda a, b: a + b
print(add(3, 4))

numbers = [1, 2, 3, 4, 5, 6]
print(list(map(lambda x: x * 10, numbers)))          # transform
print(list(filter(lambda x: x % 2 == 0, numbers)))   # filter

students = [("Ann", 80), ("Ben", 65), ("Cathy", 92)]
print(sorted(students, key=lambda s: s[1], reverse=True))  # sort by score
```

```python
# ============ TOPIC 3A: LISTS (ordered, changeable) ============
marks = [55, 78, 62, 91]
marks.append(84)            # add to end
marks.insert(1, 70)         # insert at position 1
marks.remove(62)            # remove a value
print(marks)
print("First:", marks[0], "| Last:", marks[-1], "| Slice:", marks[1:3])
marks.sort()
print("Sorted:", marks, "| Length:", len(marks))

# List comprehension
squares = [x ** 2 for x in range(1, 6)]
print(squares)

# ============ TOPIC 3B: TUPLES (ordered, NOT changeable) ============
point = (4, 7)
print(point[0], point[1])
x, y = point                # unpacking
print("x =", x, "y =", y)
# point[0] = 10   # <- would raise TypeError (tuples are immutable)
```

```python
# ============ TOPIC 4B: EXCEPTION HANDLING ============

# ---------- Basic try / except ----------
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")

# ---------- Multiple exceptions + else + finally ----------
def safe_divide(a, b):
    try:
        answer = a / b
    except ZeroDivisionError:
        print("Error: division by zero")
    except TypeError:
        print("Error: both values must be numbers")
    else:
        print("Result:", answer)       # runs only if no error
    finally:
        print("Done.")                 # always runs

safe_divide(10, 2)
safe_divide(10, 0)
safe_divide(10, "a")

# ---------- Handling a missing file ----------
try:
    with open("missing_file.txt") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found, please check the file name")

# ---------- Raising your own exception ----------
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

try:
    set_age(-5)
except ValueError as e:
    print("Caught:", e)
```

```python
# ============ TOPIC 5A: CLASSES AND OBJECTS ============
class Student:
    school = "Victoria University"       # class attribute

    def __init__(self, name, reg_no):
        self.name = name                  # instance attributes
        self.reg_no = reg_no

    def introduce(self):
        return f"I am {self.name} ({self.reg_no}) at {Student.school}"

s1 = Student("Leticia", "VU001")
s2 = Student("Namutebi", "VU002")
print(s1.introduce())
print(s2.introduce())

# ============ TOPIC 5B: ENCAPSULATION ============
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance          # private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid or insufficient funds")

    def get_balance(self):
        return self.__balance

acc = BankAccount("Leticia", 1000)
acc.deposit(500)
acc.withdraw(200)
print("Balance:", acc.get_balance())
# print(acc.__balance)   # <- AttributeError: data is protected
```

```python
# ============ TOPIC 5C: INHERITANCE ============
class Person:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return f"{self.name} says hello"

class Lecturer(Person):                    # child class
    def __init__(self, name, subject):
        super().__init__(name)             # call parent constructor
        self.subject = subject
    def speak(self):                       # method overriding
        return f"{self.name} teaches {self.subject}"

class Learner(Person):
    def speak(self):
        return f"{self.name} is learning Python"

# ============ TOPIC 5D: POLYMORPHISM ============
people = [Person("Guest"), Lecturer("Dr. Mirembe", "Data Science"), Learner("Leticia")]
for p in people:
    print(p.speak())      # same method name, different behaviour
```

```python
# ============ TOPIC 6A: NUMPY ============
import numpy as np

# ---------- Arrays ----------
arr = np.array([10, 20, 30, 40, 50])
print(arr, arr.shape, arr.dtype)
print("Mean:", arr.mean(), "| Std:", arr.std(), "| Sum:", arr.sum())
print(arr * 2)                 # vectorised operation, no loop needed
print(arr[arr > 25])           # boolean filtering

# ---------- Useful generators ----------
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.zeros((2, 3)))
print(np.random.randint(1, 100, size=5))

# ---------- Matrices ----------
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("A + B:\n", A + B)
print("Element-wise A * B:\n", A * B)
print("Matrix product A @ B:\n", A @ B)
print("Transpose:\n", A.T)
print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))

# ---------- Reshape and slicing ----------
m = np.arange(1, 13).reshape(3, 4)
print(m)
print("Row 1:", m[1], "| Column 2:", m[:, 2], "| Sub-matrix:\n", m[0:2, 1:3])
```
